```python
import random
import statistics
from collections import Counter

# ============================================================
# Задание 1. Генерация случайного списка
# ============================================================
def generate_random_list(a: int, b: int, length: int) -> list:
    """
    Принимает три параметра (a: int, b: int, length: int) 
    и возвращает список длины length из равномерно распределенной 
    случайной величины между числами a и b (включительно).
    """
    if a > b:
        raise ValueError("Параметр 'a' должен быть меньше или равен 'b'")
    
    return [random.randint(a, b) for _ in range(length)]

# Проверка работы
my_list = generate_random_list(1, 5, 20)
print(f"Сгенерированный список: {my_list}")


# ============================================================
# Задание 2. Частотный ряд (словарь уникальных значений)
# ============================================================
def get_frequency_dict(input_list: list) -> dict:
    """
    Принимает список и возвращает dict с количеством уникальных значений.
    Пример: для [1, 2, 3, 1, 2, 2] вернет {1: 2, 2: 3, 3: 1}
    """
    return dict(Counter(input_list))

# Проверка работы
freq_dict = get_frequency_dict(my_list)
print(f"Частотный ряд: {freq_dict}")

# Проверка на примере из задания
test_list = [1, 2, 3, 1, 2, 2]
print(f"Проверка на примере: {get_frequency_dict(test_list)}")


# ============================================================
# Задание 3. Статистические характеристики
# ============================================================
def get_statistics(input_list: list) -> dict:
    """
    Считает минимум, среднюю, медиану, моду и максимум для списка.
    Возвращает словарь с результатами.
    """
    if not input_list:
        return {}
    
    # Мода может отсутствовать, если все значения уникальны
    try:
        mode_val = statistics.mode(input_list)
    except statistics.StatisticsError:
        mode_val = "Нет моды (все значения уникальны)"

    return {
        "min": min(input_list),
        "mean": statistics.mean(input_list),
        "median": statistics.median(input_list),
        "mode": mode_val,
        "max": max(input_list)
    }

# Проверка работы
stats = get_statistics(my_list)
print("Статистика:")
for key, value in stats.items():
    print(f"  {key}: {value}")
    
```
